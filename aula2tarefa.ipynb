```python
import numpy as np

classes = ["pet", "tecnologia", "alimentacao"]

embeddings = {
    "gato": np.array([0.9, 0.0, 0.1]),
    "cachorro": np.array([0.9, 0.0, 0.1]),
    "robo": np.array([0.1, 0.9, 0.0]),
    "computador": np.array([0.0, 0.9, 0.0]),
    "dados": np.array([0.0, 0.8, 0.1]),
    "algoritmo": np.array([0.0, 0.95, 0.0]),
    "pizza": np.array([0.0, 0.0, 0.90]),
    "hamburguer": np.array([0.0, 0.0, 0.90]),
    "carne": np.array([0.0, 0.0, 0.95]),
    "comida": np.array([0.0, 0.0, 0.90]),
    "bife": np.array([0.0, 0.0, 0.95])
}

pergunta = "Eu gosto de pizza, hamburguer, carne e comida"
tokens = pergunta.lower().replace(",", "").split()

print(tokens)
```

```python
pesos = {
    "pizza": 2,
    "hamburguer": 2,
    "carne": 2,
    "comida": 2,
    "bife": 2
}

vetores = []

for token in tokens:
    if token in embeddings:
        peso = pesos.get(token, 1)
        for _ in range(peso):
            vetores.append(embeddings[token])

vetor_medio = np.mean(vetores, axis=0)
indice = np.argmax(vetor_medio)
tema = classes[indice]

print("Tema identificado:", tema)
print("Vetor médio:", vetor_medio)
```

```python
pesos = {
    "pizza": 0.5,
    "hamburguer": 0.5,
    "carne": 0.5,
    "comida": 0.5,
    "bife": 0.5
}

vetores = []

for token in tokens:
    if token in embeddings:
        peso = pesos.get(token, 1)
        vetores.append(embeddings[token] * peso)

vetor_medio = np.mean(vetores, axis=0)
indice = np.argmax(vetor_medio)
tema = classes[indice]

print("Tema identificado:", tema)
print("Vetor médio:", vetor_medio)
```
