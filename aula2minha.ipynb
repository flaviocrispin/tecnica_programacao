```python
classes = ["pet", "tecnologia", "alimentação"]
```

```python
classes
```

```python
embeddings = {
    # Pet
    "gato":         np.array([0.9, 0.0, 0.1]),
    "cachorro":     np.array([0.9, 0.0, 0.1]),
    "felino":       np.array([0.9, 0.0, 0.0]),
    
    # Tecnologia
    "robo":         np.array([0.1, 0.9, 0,0]),
    "computador":   np.array([0.0, 0.9, 0.0]),
    "dados":        np.array([0.0, 0.8, 0.1]),
    "algoritmo":    np.array([0.0, 0.95, 0.0]),

    # Alimentação
    "ração":        np.array([0.6, 0.0, 0.8]),
    "carne":        np.array([0.0, 0.0, 0.9]),
    "comida":       np.array([0.1, 0.0, 0.9]),
    "bife":         np.array([0.0, 0.0, 0.95]),
}
```

```python
pergunta = "o cachorro é feliz"
```

```python
tokens = pergunta.lower().split()
```

```python
tokens
```

```python
vetores = [embeddings[token] for token in tokens if token in embeddings]
```

```python
vetores
```

```python
palavras_encontradas = [token for token in tokens if token in embeddings]
```

```python

palavras_encontradas
```

```python
print (f"A pergunta '{pergunta}")
print (f"a palavra encontrada foi: '{palavras_encontradas}'")
print (f"O Tema da pergunta é: '{tema}'")
```

```python
classes = ["baixa prioridade", "média prioridade", "alta prioridade"]
```

```python
embeddings = {
    # baixa prioridade
    "gato":         np.array([0.9, 0.0, 0.1]),
    "cachorro":     np.array([0.9, 0.0, 0.1]),
    "felino":       np.array([0.9, 0.0, 0.0]),
    "pássaro":      np.array([0.8, 0.0, 0.3]),
    
    # média prioridade
    "ruim":         np.array([0.1, 0.9, 0.0]),
    "computador":   np.array([0.0, 0.9, 0.0]),
    "dados":        np.array([0.0, 0.8, 0.1]),
    "algoritmo":    np.array([0.0, 0.9, 0.0]),

    # alta prioridade
    "windows":      np.array([0.6, 0.0, 0.8]),
    "carne":        np.array([0.0, 0.0, 0.9]),
    "comida":       np.array([0.1, 0.0, 0.9]),
    "bife":         np.array([0.0, 0.0, 0.9]),
}
```

```python
pergunta = "Meu windows está ruim"
```

```python
tokens = pergunta.lower().split()
vetores = [embeddings[token] for token in tokens if token in embeddings]
palavras_encontradas = [token for token in tokens if token in embeddings]
palavras_encontradas
```

```python
indice_vencedor = np.argmax(np.mean(vetores, axis=0))
```

```python
tema = classes[indice_vencedor]
```

```python
print (f"A pergunta '{pergunta}")
print (f"a palavra encontrada foi: '{palavras_encontradas}'")
print (f"A prioridade é: '{tema}'")
```
