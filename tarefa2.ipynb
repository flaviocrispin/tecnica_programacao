```python
pesos = {
    "racao": 2,
    "carne": 2,
    "comida": 2,
    "bife": 2
}
```

```python
pergunta = "racao carne comida bife"  # ou: pergunta = input("Digite a pergunta: ")
tokens = pergunta.lower().replace(",", "").split()

vetores = []

for token in tokens:
    if token in embeddings:
        peso = pesos.get(token, 1)
        vetores.extend([embeddings[token]] * peso)

vetor_media = np.mean(vetores, axis=0)

indice_vencedor = np.argmax(vetor_media)
tema = classes[indice_vencedor]

print(f"Tema identificado: {tema}")
```
