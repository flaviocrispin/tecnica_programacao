```python
import numpy as np
```

```python
vetor = np.array([1, 2, 3, 4, 5])
```

```python
prioridade = ["baixa", "media", "alta"]
```

```python
embeddings = {
    # Baixa Prioridade
    "feio":          np.array([0.91, 0.51, 0.01]),
    "demora":        np.array([0.92, 0.52, 0.02]),
    "desalinhado":   np.array([0.93, 0.53, 0.03]),
    
    # Media Prioridade
    "reiniciar":     np.array([0.01, 0.91, 0.51]),
    "update":        np.array([0.02, 0.92, 0.52]),
    "senha":         np.array([0.03, 0.93, 0.53]),
    "instabilidade": np.array([0.04, 0.94, 0.54]),
    
    # Alta Prioridade
    "travado":       np.array([0.51, 0.01, 0.91]),
    "corrompido":    np.array([0.52, 0.02, 0.92]),
    "pagamento":     np.array([0.53, 0.03, 0.93]),
    "indisponível":  np.array([0.54, 0.04, 0.94]),
}
```

```python
chamado = "Não consigo alterar minha senha"
```

```python
tokens = chamado.lower().split()
```

```python
tokens
```

```python
vetores = [embeddings[token] for token in tokens if token in embeddings]
```

```python
vetores
```

```python
termos_chaves = [token for token in tokens if token in embeddings]
```

```python
termos_chaves
```

```python
vetor_media = np.mean(vetores, axis=0)
```

```python
vetor_media
```

```python
nivel = np.argmax(vetor_media)
```

```python
nivel
```

```python
tema = prioridade[nivel]
tema
```

```python
print (f"O chamado '{chamado}")
print (f"por contém os termos: '{termos_chaves}'")
print (f"foi classificado como de: '{tema}' prioridade")
```
