```python
import numpy as np
```

```python
classes = ["Contabilidade", "tecnologia", "rh"]
```

```python
embeddings = {
    # Contabilidade
    "contrato":        np.array([0.8, 0.0, 0.2]),
    "conta":    np.array([0.9, 0.0, 0.1]),
    "documento":      np.array([0.9, 0.0, 0.2]),
    
    # Tecnologia
    "robo":        np.array([0.1, 0.9, 0.0]),
    "computador":  np.array([0.0, 0.9, 0.0]),
    "dados":       np.array([0.0, 0.8, 0.1]),
    "algoritmo":   np.array([0.0, 0.95, 0.0]),
    
    # RH
    "ferias":            np.array([0.3, 0.0, 0.8]),
    "funcionario":       np.array([0.0, 0.0, 0.9]),
    "comissao":          np.array([0.5, 0.0, 0.5]),
    
}
```

```python
Pergunta = " O documento de Contabilidade é mais parecido com qual classe?"
```

```python
tokens = Pergunta.lower().split()
```

```python
tokens
```

```python
vetores = [embeddings[token] for token in tokens if token in embeddings]
```

```python
vetores
```

```python
palavras_encontradas = [token for token in tokens if token in embeddings]
```

```python
palavras_encontradas
```

```python
vetor_media = np.mean(vetores, axis=0)
```

```python
vetor_media
```

```python
indice_vencedor = np.argmax(vetor_media)
```

```python
indice_vencedor
```

```python
tema = classes[indice_vencedor]
tema
```

```python
print (f"A pergunta: '{Pergunta}' está relacionada com o tema: '{tema}'")
print (f"Palavras encontradas: {palavras_encontradas}")
print (f" o Tema da pergunta é: {tema}")
```
