**1-Carregue um csv em formato dataframe.**

```python
df = pd.read_csv("DimCustomer.csv")

print("=" * 50)
print("DATASET CARREGADO COM SUCESSO")
print("=" * 50)

print(f"\nArquivo: DimCustomer.csv")
print(f"Linhas: {df.shape[0]:,}")
print(f"Colunas: {df.shape[1]:,}")

print("\nPrimeiras 5 linhas:")
print(df.head().to_string(index=False))
```

**2- Veja as cinco primeiras e cinco últimas linhas.**

```python
print("\n--- Primeiras 5 linhas ---")
print(df.head().to_string(index=False))

print("\n--- Últimas 5 linhas ---")
print(df.tail().to_string(index=False))
```

**3- Verifique o nome das colunas, quantidades de linhas e colunas.**

```python
print("=" * 50)
print("ESTRUTURA DO DATASET")
print("=" * 50)

print(f"\nQuantidade de linhas: {df.shape[0]:,}")
print(f"Quantidade de colunas: {df.shape[1]:,}")

print("\nColunas disponíveis:")
for i, coluna in enumerate(df.columns, start=1):
    print(f"{i}. {coluna}")
```

**4- Filtre por salário anual maior que 150.000**

```python
df_filtrado = df[df["YearlyIncome"] > 150000]

print(f"\nTotal de clientes com YearlyIncome > 150.000: {len(df_filtrado)}")

print("\nPrincipais dados:")
print(
    df_filtrado[["CustomerKey", "YearlyIncome", "FirstName","LastName"]]
    .sort_values("YearlyIncome", ascending=True)
    .to_string(index=False)
)
```

**5-Calcule média da coluna YearlyIncome**

```python
media = df["YearlyIncome"].mean()
print(f"Média de YearlyIncome: {media:.2f}")
```
