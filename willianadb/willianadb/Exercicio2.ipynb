```python
classes = ["baixa", "media", "alta"]
```

```python
classes[0]
```

```python
embeddings = {
    # baixa
    "lenta":        np.array([0.6, 0.3, 0.1]),
    "impressora":    np.array([0.8, 0.1, 0.1]),
    "sugestao":      np.array([0.9, 0.0, 0.0]),
    
    # media
    "rede":        np.array([0.1, 0.6, 0.2]),
    "senha":  np.array([0.1, 0.8, 0.1]),
    "instalar":       np.array([0.0, 0.9, 0.1]),
    "email":   np.array([0.1, 0.8, 0.1]),
    
    # alta
    "urgente":       np.array([0.0, 0.0, 1.0]),
    "travado":       np.array([0.0, 0.1, 0.9]),
    "virus":      np.array([0.0, 0.1, 0.9]),
    "hackeado":        np.array([0.0, 0.05, 0.95]),
}
```

```python
pergunta = "Preciso imprimir um arquivo, mas a impressora está lenta O que devo fazer?"
```

```python
tokens = pergunta.lower().split()
```

```python
tokens
```

```python
vetores = [embeddings[token] for token in tokens if token in embeddings]
```

```python
vetores
```

```python
palavras_encontradas = [token for token in tokens if token in embeddings]
```

```python
palavras_encontradas
```

```python
vetor_media = np.mean(vetores, axis=0)
```

```python
vetor_media
```

```python
indice_vencedor = np.argmax(vetor_media)
```

```python
indice_vencedor
```

```python
tema = classes[indice_vencedor]
tema
```

```python
print (f"A pergunta '{pergunta}")
print (f"a palavra encontrada foi: '{palavras_encontradas}'")
print (f"A prioridade é: '{tema}'")
```
