```python
import numpy as np 
```

```python
texto = ["baixa prioridade", "media prioridade", "alta prioridade"]
```

```python
texto[0]
```

```python
embeddings = {
    # baixa prioridade
    "bem":        np.array([0.9, 0.0, 0.1]),
    "voando":     np.array([0.9, 0.0, 0.1]),
    "felino":     np.array([0.9, 0.0, 0.0]),
    
    # media prioridade
    "normal":      np.array([0.1, 0.9, 0.0]),
    "computador":  np.array([0.0, 0.9, 0.0]),
    "dados":       np.array([0.0, 0.8, 0.1]),
    "algoritmo":   np.array([0.0, 0.95, 0.0]),
    
    # alta prioridade
    "ruim":        np.array([0.6, 0.0, 0.8]),
    "carne":       np.array([0.0, 0.0, 0.9]),
    "comida":      np.array([0.1, 0.0, 0.9]),
    "bife":        np.array([0.0, 0.0, 0.95]),
}
```

```python
frase = "Meu Windows está ruim"
```

```python
tokens = frase.lower().split()
```

```python
tokens
```

```python
vetores = [embeddings[tokens] for tokens in tokens if tokens in embeddings]
```

```python
vetores
```

```python
palavras_encontradas = [token for token in tokens if token in embeddings]
```

```python
palavras_encontradas
```

```python
indice_venc = np.argmax(np.mean(vetores, axis=0))           
```

```python
indice_venc
```

```python
tema = texto[indice_venc]
tema
```

```python
print(f"A frase'{frase}' tem como tema: '{tema}'")
```
