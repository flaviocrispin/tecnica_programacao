```python
import numpy as np
```

```python
categorias = ["Alta prioridade", "Média prioridade", "Baixa prioridade"]
```

```python
embeddings = {
    # Alta prioridade
    "quebrado":        np.array([0.9, 0.0, 0.1]),
    "fumaça":    np.array([0.9, 0.0, 0.1]),
    "parado":      np.array([0.9, 0.0, 0.0]),
    "vírus":       np.array([0.8, 0.0, 0.2]),
    
    # Média prioridade
    "lento":        np.array([0.1, 0.9, 0.0]),
    "lentidão":  np.array([0.0, 0.9, 0.0]),
    "apagado":       np.array([0.0, 0.8, 0.1]),
    "devagar":   np.array([0.0, 0.95, 0.0]),
    "impressora": np.array([0.0, 0.9, 0.1]),
    
    # Baixa prioridade
    "duvida":       np.array([0.1, 0.2, 0.8]),
    "tela":       np.array([0.1, 0.3, 0.9]),
    "cabo":      np.array([0.1, 0.0, 0.9]),
    "mouse":        np.array([0.0, 0.2, 0.95]),
    "resolução":    np.array([0.0, 0.1, 0.9]),
    "fonte":      np.array([0.0, 0.4, 0.9]),

    "____": np.array([0.0, 0.0, 0.9])
}
```

```python
frase=input("Digite a descrição do problema: ")
```

```python
#frase = "O sistema está lento e quebrado , tenho dúvida se vai funcionar."
```

```python
tokens = frase.lower().split()
print(tokens)
```

```python
vetores = [embeddings[token] for token in tokens if token in embeddings]
```

```python
palavras_encontradas = [token for token in tokens if token in embeddings]
if len(palavras_encontradas)>0:
    vetor_media = np.mean(vetores, axis=0)
    print(vetor_media)
    print(palavras_encontradas)
else:
    palavras_encontradas = ["____"]
```

```python
indice_vencedor = np.argmax(vetor_media)
print(indice_vencedor)
print(categorias[indice_vencedor])
```

```python
categoria_vencedora = categorias[indice_vencedor]
print(categoria_vencedora)
```

```python
print (f"Descrição do problema: '{frase}'")
print (f"Palavra(s) encontrada(s): '{palavras_encontradas}'\n")
print (f"Classificação do problema: '{categoria_vencedora}'")
```
